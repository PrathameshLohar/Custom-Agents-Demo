```python
from smolagents import CodeAgent, TransformersModel

# 1. Load the model from your local directory
# This assumes the folder contains config.json and weights
model = TransformersModel(
    model_id="./local_smollm_360m",
    device_map="auto",  # Uses GPU if found, otherwise uses CPU
    max_new_tokens=1000
)
```

```python
# 2. Create the CodeAgent
# We enable 'add_base_tools' so it can use standard Python functions
agent = CodeAgent(
    tools=[], 
    model=model,
    add_base_tools=True
)
```

```python
# 3. Run the agent
# Since it is a 360M model, keep prompts straightforward for best results
prompt = "Explain what a black hole is in three simple sentences."
result = agent.run(prompt)

print("\n--- Final Agent Response ---")
print(result)
```
