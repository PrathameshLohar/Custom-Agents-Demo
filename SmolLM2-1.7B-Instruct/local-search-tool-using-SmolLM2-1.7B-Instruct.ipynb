### <b>Initializing SmolLM2 Agent </b>
Below is a basic implementation using local transformers and a simple search tool.

```python
from langchain_huggingface import HuggingFacePipeline
from langchain.agents import initialize_agent, AgentType
import DuckDuckGoSearchRun
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
```

```python
# 1. Setup local model
model_id = "HuggingFaceTB/SmolLM2-1.7B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto")

pipe = pipeline("text-generation", model=model, tokenizer=tokenizer, max_new_tokens=512)
llm = HuggingFacePipeline(pipeline=pipe)
```

```python
# 2. Define tools
tools = [DuckDuckGoSearchRun()]
```

```python
# 3. Initialize the agent
# Using ZERO_SHOT_REACT_DESCRIPTION as it's most compatible with smaller instruct models
agent = initialize_agent(
    tools, 
    llm, 
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION, 
    verbose=True,
    handle_parsing_errors=True
)
```

```python
# 4. Run the agent
agent.run("What is the current temperature in London?")
```
